SALES DATA

```python
import numpy as np
import matplotlib.pyplot as plt

months=np.arange(1,13)
print(months)
print()

month_names =np.array([
    "Jan","Feb","Mar","Apr",
    "May","Jun","Jul","Aug",
    "Sep","Oct","Nov","Dec"
])
print(month_names)
print()

stores=np.arange(1, 11)
print(stores)
print()

np.random.seed(42)
sales = np.random.randint(20000,80001,size=(10,12))
print(sales)
```

QUARTERLY SALES ,AVERAGE AND HIGHEST SALES

```python
print("Quarterly sales")
a=sales[:, 0:3]
b=sales[:, 3:6]
c=sales[:, 6:9]
d=sales[:, 9:12]

print("JANUARY-MARCH")
print(a)
print("TOTAL",a.sum(axis=1))

print("APRIL-JUNE")
print(b)
print("TOTAL",b.sum(axis=1))

print("JULY-SEPTEMBER")
print(c)
print("TOTAL",c.sum(axis=1))

print("OCTOBER-DECEMBER")
print(d)
print("TOTAL",d.sum(axis=1))

print("Annual sales")
print(sales.sum(axis=1))

print("Average annual sales of each store")
print(sales.mean(axis=1))

print("Average of total sales of all stores")
avg_sales=sales.mean()
print(avg_sales)
```

SALES MORE THAN TARGET

```python
target=60000
up_sales=sales
up_sales = np.where(up_sales>target,up_sales,0)
print(up_sales)
```

SALES BELOW AVERAGE

```python
avg_each=sales.mean(axis=1).reshape(10,1)
print(avg_each)

print()
print(avg_sales)
print()

below_avg = np.where(avg_each < avg_sales, 0, avg_each)
print(below_avg)
```

BEST 3 MONTHS OF EACH STORES

```python
store_id = [1, 4, 7]

selected_stores = sales[store_id]
print(selected_stores)

print()

monthly_sales = sales.sum(axis=0)
print(monthly_sales)

print()

best_months = np.argsort(monthly_sales)[-3:]

print("Best 3 months:", best_months + 1)
print("Sales:", monthly_sales[best_months])
```

MONTHLY GROWTH REPORT

```python
growth_targets = np.linspace(5,15,12)
target_sales = monthly_sales *(1+growth_targets/100)

print("MONTHLY GROWTH REPORT")
for i in range(12):
    print(
        "Month:", i + 1,
        "   Current Sales:", monthly_sales[i],
        "   Growth:", round(growth_targets[i], 2), "%",
        "   Target Sales:", round(target_sales[i], 2),
        "   Extra Sales Needed:", round(target_sales[i] - monthly_sales[i], 2)
    )
```

STORE WISE GROWTH REPORT

```python
growth_targets = np.linspace(5, 15, 10)

store_sales = sales.sum(axis=1)
target_sales = store_sales * (1 + growth_targets/100)
print("STORE-WISE GROWTH REPORT")

for i in range(10):
    print(
        "Store:", i + 1,
        "   Current Sales:", store_sales[i],
        "   Growth:", round(growth_targets[i], 2), "%",
        "   Target Sales:", round(target_sales[i], 2),
        "   Extra Sales Needed:", round(target_sales[i] - store_sales[i], 2)
    )
```
